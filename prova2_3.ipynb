```python
import math
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.optimize import curve_fit
from scipy.stats import norm
from scipy.stats import chi2

# Stile comune con accento rosso del corso.
ROSSO = "#C92512"
GRIGIO = "#5A5A5A"
AZZURRO = "#2E6F9E"
ORO = "#D79B00"
plt.rcParams.update({
    "figure.figsize": (8.2, 4.6),
    "figure.dpi": 120,
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.labelsize": 11,
    "axes.titlesize": 12,
    "font.size": 11,
    "legend.frameon": False,
})
```

```python
# Caricamento del file CSV
dati = np.genfromtxt(
    "dati_modello.csv",
    delimiter=",",
    names=True
)

# Estrazione delle colonne
x = dati["x_mm"]
y = dati["conteggi"]
sigma_y = dati["sigma_y"]

# Verifica lunghezze
if len(x)== len(y) == len(sigma_y):
   print("Le lunghezze degli array sono uguali")
else:
   print("Gli array non hanno la stessa lunghezza")

# Verifica valori finiti
if np.all(np.isfinite(x)):
   print("Tutti i valori di x sono finiti")
else:
   print("Non tutti i valori di x sono finiti")

if np.all(np.isfinite(y)):
   print("Tutti i valori di y sono finiti")
else:
   print("Non tutti i valori di y sono finiti")

if np.all(np.isfinite(sigma_y)):
   print("Tutti i valori di sigma_y sono finiti")
else:
   print("Non tutti i valori di sigma_y sono finiti")


# Verifica incertezze positive
if np.all(sigma_y > 0):
   print("Tutte le incertezze sono positive")
else:
   print("Almeno un'incertezza non è positiva")

# Verifica posizioni ordinate
if np.all(np.diff(x) > 0):
   print("Le posizioni sono ordinate")
else:
   print("Le posizioni non sono strettamente crescenti")

# Intervalli dei dati
print(f"Numero di punti:", len(x))
print(f"Intervallo x: [{x.min():.3f}, {x.max():.3f}] mm")
print(f"Intervallo y: [{y.min():.3f}, {y.max():.3f}] conteggi")
print(f"Intervallo sigma_y: [{sigma_y.min():.3f}, {sigma_y.max():.3f}] conteggi")
```

```python
#Rappresentazione dati con barre d'errore
fig, ax = plt.subplots()

ax.errorbar(
    x, y, 
    yerr=sigma_y,
    fmt='o',
    markersize=3,
    capsize=2,
    color=AZZURRO,
    label="Dati con barre d'errore"
)

ax.set(xlabel="Posizione x (mm)", ylabel="Conteggi per bin", title="Dati con barre d'errore")
ax.legend()
plt.tight_layout()
plt.show()


# Selezione delle regioni laterali
maschera_laterale = (x < -4) | (x > 4)
# Stima del fondo come mediana dei conteggi selezionati
B_stimato = np.median(y[maschera_laterale])
print(f"Fondo stimato: {B_stimato:.2f} conteggi/bin")

# Massimo centrale
maschera_centrale = np.abs(x) < 0.3
massimo_centrale = np.max(y[maschera_centrale])
print("Massimo centrale:", massimo_centrale )


#distanza tra le frangie stimata come ampiezza intervallo/numero di minimi
distanza = 4/8  
print(f"Distanza tra due picchi:", distanza)          

# primo minimo dell'inviluppo
x_min=3.5 
print(f"Primo minimo dell'inviluppo in x =", x_min)
```

```python
# Stime iniziali del fondo e dell'ampiezza
B_0 = B_stimato
I_0 = massimo_centrale-B_0

# stime iniziali dei parametri ricavati dal grafico
alpha_0 = np.pi/x_min          
delta_0 = np.pi/distanza

# vettore stime iniziali
p0 = [B_0, I_0, alpha_0, delta_0]

print(f"Stime iniziali dei parametri: \nB_0 = {p0[0]}, I_0 = {p0[1]}, alpha_0 = {p0[2]:.3f}, delta_0 = {p0[3]:.3f}")

# impostare i bounds
lower = [0.0, 0.0, 0.10, 1.2]
upper = [np.inf, np.inf, 3.0, 12.0]

```

```python
# Definizione funzione del modello
def E(x, alpha): 
    return np.sinc(alpha * x / np.pi) **2

def C(x, delta):
    return np.cos(delta * x) **2


#funzione principale
def mu(x, B, I, alpha, delta):
    return B + I * E(x,alpha) * C(x,delta)

# Esecuzione del fit
popt, pcov = curve_fit(
    mu, x, y,
    p0=p0,
    sigma=sigma_y,
    absolute_sigma=True,
    bounds=(lower,upper),
    maxfev=100000
)

# Parametri ottimali
B_fit, I_fit, alpha_fit, delta_fit = popt

# Incertezze sui parametri
incertezze_standard = np.sqrt(np.diag(pcov))

# Stampa dei parametri
print("Parametri ottimali:")
print(f"B = {B_fit:.1f} ± {incertezze_standard[0]:.1f} conteggi")
print(f"I = {I_fit:.0f} ± {incertezze_standard[1]:.0f} conteggi")
print(f"alpha = {alpha_fit:.3f} ± {incertezze_standard[2]:.3f} 1/mm")
print(f"delta = {delta_fit:.3f} ± {incertezze_standard[3]:.3f} 1/mm")

def covarianza_correlazione(cov):
    deviazioni = np.sqrt(np.diag(cov))
    return cov / np.outer(deviazioni, deviazioni)

# Matrice di correlazione
corr = covarianza_correlazione(pcov)

with np.printoptions(precision=5, suppress=True):
    print("Matrice di correlazione:")
    print(corr)

# Mappa di colore
plt.imshow(corr, vmin=-1, vmax=1, cmap="RdBu_r")
plt.colorbar(label="Correlazione")
plt.xticks(range(4), ["B", "I", "alpha", "delta"])
plt.yticks(range(4), ["B", "I", "alpha", "delta"])
plt.title("Matrice di correlazione")
plt.tight_layout()
plt.show()


# La coppia maggiormente correlata è costituita da I e alpha, 
# con coefficiente di correlazione 0.434, che indica una correlazione positiva moderata.
# I due parametri tendono a variare nella stessa direzione e i loro effetti sulla curva 
# possono essere parzialmente correlati.
```

```python
# Valori previsti dal modello
y_fit = mu(x, *popt)

# Residui e pull
residui = y - y_fit
pull = residui / sigma_y

# Gradi di libertà
N = len(y)
ndf = N - 4

# chi-quadro e chi-quadro ridotto
chi2_obs = np.sum(pull**2)
chi2_ridotto = chi2_obs / ndf

# p-value
p_value=chi2.sf(chi2_obs, ndf)

# costruzione figura
fig, ax = plt.subplots(
    2,1, figsize=(8.2,6.4), sharex=True,
    gridspec_kw={"height_ratios":[2.1,1]}
)

ax[0].errorbar(
    x, y,
    yerr=sigma_y,
    fmt='o',
    color=AZZURRO,
    markersize=2.5,
    capsize=2.5,
    label="Punti sperimentali"
)
ax[0].set(xlabel=" x (mm)", ylabel="mu(x)",  title="Fit e diagnostica dei residui")
ax[0].plot(x, y_fit, color=ROSSO, lw=2)

ax[1].axhline(0, color=GRIGIO, lw=1.4)
ax[1].axhspan(-1,1, color=ORO, alpha=0.16)
ax[1].plot(x, pull, 'o', markersize=2, color=ROSSO)
ax[1].set(xlabel="x (mm)", ylabel="pull")
fig.tight_layout()
plt.show()


```

```python
# Fattori di calibrazione e rispettive incertezze
k = 1.000
sx = 1.000

u_k = 0.010
u_sx = 0.005

# Vettore dei parametri
z = np.array([alpha_fit, delta_fit, k, sx])
A = np.column_stack(z)
print(f"Vettore dei parametri:\n {A.T}")

C_alpha_delta = pcov[2:4,2:4]             #Blocco di covarianza di alpha e delta ricavato dal fit

# Costruzione della matrice di covarianza 
C_z = np.zeros((4,4)) 
C_z[:2, :2]= C_alpha_delta                
C_z[2,2]=u_k**2                           #Inseriamo sulla diagonale di C_z le varizanze di k e sx
C_z[3,3]=u_sx**2

print("Matrice di covarianza C_z:")
print(C_z)

# Controllo simmetria, diagonale positiva, autovalori non negativi
if np.allclose(C_z, C_z.T):
   print("C_z è simmetrica")
else:
    print("C_z non è simmetrica")

if np.all(np.diag(C_z)>0):
    print("C_z ha diagonale positiva")
else:
    print("C_z non ha diagonale tutta positiva")

autovalori = np.linalg.eigvalsh(C_z)

if np.all(autovalori>=0):
    print("C_z ha autovalori non negativi")
else:
    print("C_z ha almeno un autovalore negativo")


# Matrice di correlazione
C_corr = covarianza_correlazione(C_z)

with np.printoptions(precision=5, suppress=True):
    print("Matrice di correlazione di alpha, delta, k, sx:")
    print(C_corr)

# Mappa di colore
plt.imshow(C_corr, vmin=-1, vmax=1, cmap="coolwarm")
plt.colorbar(label="Correlazione")
plt.xticks(range(4), ["alpha", "delta", "k", "sx"])
plt.yticks(range(4), ["alpha", "delta", "k", "sx"])

# Valori numerici nelle celle
for i in range(4):
    for j in range(4):
        plt.text(j, i, f"{C_corr[i,j]:.3f}",
                 ha="center", va="center", color="black")

plt.title("Matrice di correlazione")
plt.tight_layout()
plt.show()

# Controllo che la matrice di correlazione sia simmetrica e con diagonale positiva
if np.allclose(C_corr, C_corr.T):
    print("La matrice di correlazione C_corr è simmetrica")
else:
    print("C_corr non è simmetrica")
if np.all(np.diag(C_corr)>0):
    print("C_corr ha diagonale positiva")
else:
    print("C_corr non ha diagonale strettamente positiva")
```

```python
a = alpha_fit/(k*sx)
d = delta_fit/(k*sx)
q = d/a

g = np.array([a, d, q])
g_z = np.column_stack(g)

# Matrice jacobiana
J = np.array([
    [1/(k*sx), 0, -alpha_fit/(k**2*sx), -alpha_fit/(k*sx**2)],

    [0, 1/(k*sx), -delta_fit/(k**2*sx), -delta_fit/(k*sx**2)],

    [-delta_fit/alpha_fit**2, 1/alpha_fit, 0, 0]
])

print("Matrice Jacobiana:")
print(J)

# matrice di covarianza di a, d, q
C_g = J @ C_z @ J.T

print("Matrice di covarianza di g:")
print(C_g)

# Valori centrali
print("Valori centrali:")
print(f"a = {a:.3f} 1/mm")
print(f"d = {d:.2f}  1/mm")
print(f"q = {q:.2f}")

# Incertezze marginali
incertezze = np.sqrt(np.diag(C_g))
u_a = incertezze[0]
u_d = incertezze[1]
u_q = incertezze[2]
print("Incertezze marginali:")
print(f"u_a = {u_a:.3f} 1/mm")
print(f"u_d = {u_d:.2f}  1/mm")
print(f"u_q = {u_q:.2f}")

#Matrice di correlazione
C_g_corr = covarianza_correlazione(C_g)

print("Matrice di correlazione:")
with np.printoptions(precision=5, suppress=True):
    print(C_g_corr)


#Mappa di colore matrice di correlazione a, d, q 
plt.figure(figsize=(6,5))

plt.imshow(C_g_corr, vmin=-1, vmax=1, cmap="coolwarm")
plt.colorbar(label="Correlazione")
plt.xticks(range(3), ["a", "d", "q"])
plt.yticks(range(3), ["a", "d", "q"])

for i in range(3):
    for j in range(3):
        plt.text(j, i, f"{C_g_corr[i,j]:.3f}",
                 ha="center", va="center")

plt.title("Matrice di correlazione di a, d, q")
plt.tight_layout()
plt.show()

# Verifica per q
sinistra = (u_q/q)**2
destra = (incertezze_standard[3]/delta_fit)**2 + (incertezze_standard[2]/alpha_fit)**2 - 2*C_alpha_delta[0,1]/(alpha_fit*delta_fit)
print(f"(u_q/q)^2 = {sinistra}")
print(f"propagazione analitica = {destra}")
if sinistra == destra:
    print(f"Verifica per q: verificato :)")
else:
    print("verifica per q: non verificato :(")
```

```python
#Propagazione Monte Carlo correlata
M = 100_000

media_input = z

rng = np.random.default_rng(20261003)
campioni = rng.multivariate_normal(
    mean=media_input,
    cov=C_z,
    size=M
)

# Estrazione delle colonne
alpha_mc = campioni[:, 0]
delta_mc = campioni[:, 1]
k_mc = campioni[:, 2]
sx_mc = campioni[:, 3]

# Controllo di ammissibilità
ammissibili = ( (alpha_mc > 0) & (delta_mc > 0) & (k_mc > 0) & (sx_mc > 0))
somma_ammissibili = np.sum(ammissibili)
print("Campioni ammissibili:", somma_ammissibili)
print("Campioni non ammissibili:", M-somma_ammissibili)

# Selezione dei campioni ammissibili
alpha_validi = alpha_mc[ammissibili]
delta_validi = delta_mc[ammissibili]
k_validi = k_mc[ammissibili]
sx_validi = sx_mc[ammissibili]

# Grandezze derivate
a_mc = alpha_validi / (k_validi * sx_validi)
d_mc = delta_validi / (k_validi * sx_validi)
q_mc = d_mc / a_mc

def statistiche(valori):

    media = np.mean(valori)
    std = np.std(valori, ddof=1)
    mediana = np.median(valori)

    p16 = np.percentile(valori, 16)
    p84 = np.percentile(valori, 84)

    return media, std, mediana, p16, p84

stat_a = statistiche(a_mc)
stat_d = statistiche(d_mc)
stat_q = statistiche(q_mc)

nomi = ["a", "d", "q"]
risultati = [stat_a, stat_d, stat_q]

for nome, valori in zip(nomi, risultati):

    media, std, mediana, p16, p84 = valori

    print(f"\nGrandezza {nome}:")
    print(f"Media = {media:.6f}")
    print(f"Deviazione standard = {std:.6f}")
    print(f"Mediana = {mediana:.6f}")
    print(f"Percentile 16 = {p16:.6f}")
    print(f"Percentile 84 = {p84:.6f}")

# Risultati della propagazione lineare
media_lineare = g
std_lineare = np.sqrt(np.diag(C_g))

# Risultati Monte Carlo
media_mc = np.array([
    np.mean(a_mc),
    np.mean(d_mc),
    np.mean(q_mc)
])

std_mc = np.array([
    np.std(a_mc, ddof=1),
    np.std(d_mc, ddof=1),
    np.std(q_mc, ddof=1)
])

# Tabella di confronto
print("\nCONFRONTO PROPAGAZIONE LINEARE E MONTE CARLO")

for i, nome in enumerate(["a", "d", "q"]):

    print(f"\n{nome}:")
    print(f"Media lineare: {media_lineare[i]:.6f}")
    print(f"Media Monte Carlo: {media_mc[i]:.6f}")
    print(f"Std lineare: {std_lineare[i]:.6f}")
    print(f"Std Monte Carlo: {std_mc[i]:.6f}")
```

```python
#studio blocco correlato (alpha, delta)

C_alpha_delta = pcov[2:4, 2:4]

print("Matrice di covarianza alpha-delta:")
print(C_alpha_delta)

# Coefficiente di correlazione
rho_alpha_delta = C_alpha_delta[0,1] / (
    np.sqrt(C_alpha_delta[0,0] * C_alpha_delta[1,1])
)

print("Correlazione alpha-delta =", rho_alpha_delta)

#propagazione analitica del solo fit
C_fit = np.zeros((4,4))

C_fit[:2,:2] = C_alpha_delta

C_g_fit = J @ C_fit @ J.T

u_fit = np.sqrt(np.diag(C_g_fit))

print("\nContributo del fit:")
print("u(a) =", u_fit[0])
print("u(d) =", u_fit[1])
print("u(q) =", u_fit[2])

#simulazione montecarlo dal fit
rng_fit = np.random.default_rng(20261003)

campioni_fit = rng_fit.multivariate_normal(
    mean=[alpha_fit, delta_fit],
    cov=C_alpha_delta,
    size=M
)

alpha_fit_mc = campioni_fit[:,0]
delta_fit_mc = campioni_fit[:,1]

# Controllo di ammissibilità
validi_fit = (alpha_fit_mc > 0) & (delta_fit_mc > 0)

alpha_fit_mc = alpha_fit_mc[validi_fit]
delta_fit_mc = delta_fit_mc[validi_fit]

# Grandezze derivate
a_fit_mc = alpha_fit_mc / (k*sx)
d_fit_mc = delta_fit_mc / (k*sx)
q_fit_mc = delta_fit_mc / alpha_fit_mc

print("\nDeviazioni standard Monte Carlo - FIT")

print("u(a) =", np.std(a_fit_mc, ddof=1))
print("u(d) =", np.std(d_fit_mc, ddof=1))
print("u(q) =", np.std(q_fit_mc, ddof=1))

#grafici del fit

#correlazione tra i parametri
plt.figure(figsize=(7,5))

plt.scatter(alpha_fit_mc, delta_fit_mc,
            s=8, alpha=0.3, color="steelblue")

plt.scatter(alpha_fit, delta_fit,
            color="red", marker="*", s=180,
            label="Valori nominali")

plt.xlabel(r"$\alpha$ (mm$^{-1}$)")
plt.ylabel(r"$\delta$ (mm$^{-1}$)")
plt.title("Correlazione Monte Carlo del fit")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

#distribuzione di q
plt.figure(figsize=(8,5))

plt.hist(q_fit_mc, bins=60, density=True,
         color="steelblue", alpha=0.7)

plt.axvline(q, color="red", linestyle="--",
            label="Valore nominale")

plt.axvline(np.mean(q_fit_mc), color="green",
            linestyle="-.", label="Media MC")

plt.axvline(np.median(q_fit_mc), color="orange",
            linestyle=":", label="Mediana MC")

plt.xlabel("q")
plt.ylabel("Densità")
plt.title("Distribuzione di q - solo contributo del fit")
plt.legend()
plt.grid(alpha=0.2)
plt.show()
```

```python
#studio calibrazione k

#propagazione analitica
C_k = np.zeros((4,4))
C_k[2,2] = u_k**2

C_g_k = J @ C_k @ J.T

u_k_contributo = np.sqrt(np.diag(C_g_k))

print("\nContributo della calibrazione k:")
print("u(a) =", u_k_contributo[0])
print("u(d) =", u_k_contributo[1])
print("u(q) =", u_k_contributo[2])

#propagazione monte carlo
rng_k = np.random.default_rng(20261004)

k_mc = rng_k.normal(k, u_k, M)

validi_k = k_mc > 0
k_mc = k_mc[validi_k]

a_k_mc = alpha_fit / (k_mc*sx)
d_k_mc = delta_fit / (k_mc*sx)
q_k_mc = np.full(len(k_mc), q)

print("\nDeviazioni standard Monte Carlo - k:")
print("u(a) =", np.std(a_k_mc, ddof=1))
print("u(d) =", np.std(d_k_mc, ddof=1))
print("u(q) =", np.std(q_k_mc, ddof=1))

#grafici 
fig, assi = plt.subplots(1,2, figsize=(11,4))

assi[0].hist(a_k_mc, bins=60, color="cornflowerblue",
             alpha=0.7, edgecolor="black")
assi[0].axvline(a, color="red", linestyle="--",
                label="Valore nominale")
assi[0].set_title("Distribuzione di a - calibrazione k")
assi[0].set_xlabel("a")
assi[0].legend()

assi[1].hist(d_k_mc, bins=60, color="salmon",
             alpha=0.7, edgecolor="black")
assi[1].axvline(d, color="red", linestyle="--",
                label="Valore nominale")
assi[1].set_title("Distribuzione di d - calibrazione k")
assi[1].set_xlabel("d")
assi[1].legend()

plt.tight_layout()
plt.show()
```

```python
#studio calibrazione asse sx

#propagazione analitica
C_sx = np.zeros((4,4))
C_sx[3,3] = u_sx**2

C_g_sx = J @ C_sx @ J.T

u_sx_contributo = np.sqrt(np.diag(C_g_sx))

print("\nContributo della calibrazione sx:")
print("u(a) =", u_sx_contributo[0])
print("u(d) =", u_sx_contributo[1])
print("u(q) =", u_sx_contributo[2])

#propagazione Monte Carlo
rng_sx = np.random.default_rng(20261005)

sx_mc = rng_sx.normal(sx, u_sx, M)

validi_sx = sx_mc > 0
sx_mc = sx_mc[validi_sx]

a_sx_mc = alpha_fit / (k*sx_mc)
d_sx_mc = delta_fit / (k*sx_mc)
q_sx_mc = np.full(len(sx_mc), q)

print("\nDeviazioni standard Monte Carlo - sx:")
print("u(a) =", np.std(a_sx_mc, ddof=1))
print("u(d) =", np.std(d_sx_mc, ddof=1))
print("u(q) =", np.std(q_sx_mc, ddof=1))

#grafici
fig, assi = plt.subplots(1,2, figsize=(11,4))

assi[0].hist(a_sx_mc, bins=60, color="cornflowerblue",
             alpha=0.7, edgecolor="black")
assi[0].axvline(a, color="red", linestyle="--",
                label="Valore nominale")
assi[0].set_title("Distribuzione di a - calibrazione sx")
assi[0].set_xlabel("a")
assi[0].legend()

assi[1].hist(d_sx_mc, bins=60, color="salmon",
             alpha=0.7, edgecolor="black")
assi[1].axvline(d, color="red", linestyle="--",
                label="Valore nominale")
assi[1].set_title("Distribuzione di d - calibrazione sx")
assi[1].set_xlabel("d")
assi[1].legend()

plt.tight_layout()
plt.show()
```
